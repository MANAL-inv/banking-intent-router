```python
import matplotlib.pyplot as plt
from src import config

fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
panels = [("coverage", "coverage (fraction answered)"),
          ("acc_all", "accuracy over all queries"),
          ("wrong_and_accepted", "answered AND wrong")]

for ax, (col, title) in zip(axes, panels):
    for track, grp in rob.groupby("track"):
        grp = grp.sort_values("noise_rate")
        ax.plot(grp["noise_rate"] * 100, grp[col], marker="o", label=track)
    ax.set_xlabel("character corruption rate (%)")
    ax.set_title(title)
    ax.grid(alpha=0.3)
axes[0].set_ylabel("fraction")
axes[2].legend()
plt.tight_layout()
fig.savefig(config.FIGURES_DIR / "robustness.png", dpi=150)
plt.show()
```
